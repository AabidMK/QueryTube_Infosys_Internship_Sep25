```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from langchain_huggingface import HuggingFaceEmbeddings

# Load the merged dataset
# Same logic as before: build one text field per video from title + transcript

df = pd.read_csv('../data/merged_output.csv')
df['text_for_embedding'] = df['title'].fillna('') + ' ' + df['transcript'].fillna('')

# Keep the existing sentence-transformer workflow intact
model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(df['text_for_embedding'].tolist(), show_progress_bar=True)

# LangChain-compatible embedding wrapper (no logic change to the retrieval pipeline)
langchain_embedding = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)
_ = langchain_embedding.embed_query(df['text_for_embedding'].iloc[0])

# Save embeddings as a new column
# This preserves the original output format exactly

df['embedding'] = [np.array(e).tolist() for e in embeddings]

# Save the dataset with embeddings
# Same destination and structure as before
df.to_csv('merged_with_embeddings.csv', index=False)

print("Embeddings generated and saved to merged_with_embeddings.csv")
print("LangChain embedding wrapper initialized successfully.")
```
